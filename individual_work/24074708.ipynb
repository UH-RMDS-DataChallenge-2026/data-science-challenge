```python
# ============================================================
# DATA SCIENCE CHALLENGE — TOWN A (Seasonal Flu Vaccine)
# Metric  : Total town cost (vaccine $300, hospital $18,500)
# Output  : challenge_submission_group_X_order_Y.csv
# Ranking : by CV cost (holdout used once for reporting only)
# ============================================================

# %%
!pip install -q lightgbm xgboost shap

# %%
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
import warnings, random, json, platform, sklearn
from sklearn.model_selection import (
    train_test_split, StratifiedKFold, cross_validate, RandomizedSearchCV
)
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.metrics import roc_auc_score, brier_score_loss
from sklearn.calibration import calibration_curve
from sklearn.ensemble import (
    RandomForestClassifier, ExtraTreesClassifier, HistGradientBoostingClassifier
)
from sklearn.base import clone
from sklearn.inspection import permutation_importance, PartialDependenceDisplay
from lightgbm import LGBMClassifier
from xgboost import XGBClassifier

warnings.filterwarnings('ignore')
pd.set_option('display.max_columns', 200)
sns.set_style('whitegrid')
RANDOM_STATE = 42
np.random.seed(RANDOM_STATE); random.seed(RANDOM_STATE)

print('Package versions:', json.dumps({
    'python': platform.python_version(), 'numpy': np.__version__,
    'pandas': pd.__version__, 'sklearn': sklearn.__version__,
}, indent=2))

# %% [markdown]
# ## Configuration
#
# ### Hospitalisation probabilities — justification required
#
# The slides require **P(hospitalised | infection)**. The rates below are
# **literature-informed central estimates** for seasonal influenza, treated
# as proxies under the slides' "100% infection" assumption. **The group must
# cite the exact source used in the presentation.**
#
# - Working age (18–64): central ≈ 0.5–1.0% per infection (pooled estimates,
#   e.g. Cromer et al. 2014; Tokars et al. 2018). Range considered: 0.25–2.0%.
# - Older adults (65+): central ≈ 2.5% per infection. Range considered: 1.0–5.0%.
#
# ### Vaccine effectiveness against hospitalisation
#
# UK SARI 2022/23 and 2023/24 report VE ≈ 34–38% (18–64) and 18–25% (65+).
# Central values: 0.35 (18–64) and 0.22 (65+). Sensitivity: ±0.10.

# %%
GROUP_NUMBER = 0        # TODO: set your group number
TRAIN_PATH   = 'dataset_A_training.csv'
TEST_PATH    = 'dataset_A_testing.csv'
SAMPLE_PATH  = 'dataset_A_example_submission.csv'
TARGET_COL   = 'seasonal_vaccine'
ID_COL       = 'respondent_id'

VACCINE_COST         = 300
HOSPITALISATION_COST = 18_500

# Age ordinal mapping (deterministic; used as the single source of truth)
AGE_MAP = {
    '18 - 34 Years': 1, '35 - 44 Years': 2, '45 - 54 Years': 3,
    '55 - 64 Years': 4, '65+ Years': 5,
}
ORD_TO_AGE = {v: k for k, v in AGE_MAP.items()}

# Central estimates
HU_CENTRAL = {1: 0.005, 2: 0.005, 3: 0.008, 4: 0.010, 5: 0.025}
VE_CENTRAL = {1: 0.35,  2: 0.35,  3: 0.35,  4: 0.35,  5: 0.22}

# Sensitivity scenarios
SENSITIVITY = {
    'low':     ({k: v*0.5 for k, v in HU_CENTRAL.items()},
                {k: max(0.10, v-0.10) for k, v in VE_CENTRAL.items()}),
    'central': (HU_CENTRAL, VE_CENTRAL),
    'high':    ({k: v*2.0 for k, v in HU_CENTRAL.items()},
                {k: min(0.50, v+0.10) for k, v in VE_CENTRAL.items()}),
}

def threshold_for_ord(ord_age, hu_dict, ve_dict):
    hu = hu_dict[ord_age]
    hv = hu * (1 - ve_dict[ord_age])
    return VACCINE_COST / (VACCINE_COST + (hu - hv) * HOSPITALISATION_COST)

print('Per-age thresholds (central):')
for o in sorted(HU_CENTRAL):
    print(f'  ord={o} ({ORD_TO_AGE[o]:15s})  t={threshold_for_ord(o, HU_CENTRAL, VE_CENTRAL):.3f}')

# %% [markdown]
# ## Cost function (single source of truth)
# Age band is derived from `age_group_ord`, which is always present in `X`.

# %%
def _get_ord_ages(X):
    if 'age_group_ord' in X.columns:
        return X['age_group_ord'].astype(float).round().astype(int).values
    raise ValueError("age_group_ord missing from X — cannot compute cost.")

def total_cost(y_true, y_score, X_reference, hu_dict=HU_CENTRAL, ve_dict=VE_CENTRAL):
    y_true = np.asarray(y_true).ravel()
    y_score = np.asarray(y_score).ravel()
    ords = _get_ord_ages(X_reference)

    thresholds = np.array([threshold_for_ord(o, hu_dict, ve_dict) for o in ords])
    send = (y_score >= thresholds).astype(int)

    hu = np.array([hu_dict[o] for o in ords])
    hv = np.array([hu_dict[o]*(1-ve_dict[o]) for o in ords])

    tp = (send == 1) & (y_true == 1)
    tn = (send == 0) & (y_true == 0)
    fp = (send == 1) & (y_true == 0)
    fn = (send == 0) & (y_true == 1)

    cost = (tp*(VACCINE_COST + hv*HOSPITALISATION_COST)
            + (fp+fn)*(VACCINE_COST + hu*HOSPITALISATION_COST)
            + tn*hu*HOSPITALISATION_COST).sum()
    return float(cost)

def negative_mean_cost(estimator, X, y):
    proba = estimator.predict_proba(X)[:, 1]
    return -total_cost(y, proba, X) / len(y)

def auc_scorer(estimator, X, y):
    return roc_auc_score(y, estimator.predict_proba(X)[:, 1])

# Sanity check
_rng = np.random.default_rng(0)
_y = _rng.integers(0, 2, 500)
_X = pd.DataFrame({'age_group_ord': _rng.integers(1, 6, 500)})
assert total_cost(_y, np.clip(_y + _rng.normal(0,0.001,500),0,1), _X) \
     < total_cost(_y, _rng.random(500), _X)
print('Cost sanity check passed.')

# %%
train  = pd.read_csv(TRAIN_PATH)
test   = pd.read_csv(TEST_PATH)
sample = pd.read_csv(SAMPLE_PATH)
assert list(sample.columns) == [ID_COL, TARGET_COL]
assert TARGET_COL not in test.columns
assert test[ID_COL].equals(sample[ID_COL]), "Test/sample ID order mismatch"
print('Train', train.shape, '| Test', test.shape)
print('Target balance:\n', train[TARGET_COL].value_counts(normalize=True).round(4))

# %% [markdown]
# ## EDA — written interpretation is added in the markdown cells below.

# %%
print('Missingness (top 20):')
display(train.isna().sum().sort_values(ascending=False).head(20))
print('Duplicate rows:', train.duplicated().sum())

fig, axes = plt.subplots(2, 3, figsize=(16, 8))
for ax, col in zip(axes.flat,
                   ['age_group','education','income_poverty',
                    'opinion_seas_vacc_effective','doctor_recc_seasonal','health_worker']):
    if col not in train.columns: continue
    if train[col].dtype == object:
        sns.countplot(data=train, x=col, hue=TARGET_COL, ax=ax); ax.tick_params(axis='x', rotation=45)
    else:
        sns.histplot(data=train, x=col, hue=TARGET_COL, multiple='stack', bins=20, ax=ax)
    ax.set_title(col)
plt.tight_layout(); plt.show()

# Interpretation prompts to paste into markdown:
# - Which categorical levels have the highest / lowest positive rate?
# - Are opinions (effectiveness, risk, sick-from-vacc) distributed bimodally?
# - Is doctor_recc_seasonal strongly associated with acceptance?

# %%
# ---------- FEATURE ENGINEERING (no blind fillna(0)) ----------
EDU_MAP = {'< 12 Years':1,'12 Years':2,'Some College':3,'College Graduate':4}
INC_MAP = {'Below Poverty':1,'<= $75,000, Above Poverty':2,'> $75,000':3}
# NOTE: census_msa is intentionally left categorical (no natural order)

def engineer(df):
    df = df.copy()
    df['age_group_ord']      = df['age_group'].map(AGE_MAP)
    df['education_ord']      = df['education'].map(EDU_MAP)
    df['income_poverty_ord'] = df['income_poverty'].map(INC_MAP)

    for c in ['health_insurance','employment_sector','income_poverty','education']:
        if c in df.columns:
            df[f'{c}_missing'] = df[c].isna().astype(int)

    behav = [c for c in df.columns if c.startswith('behavioral_')]
    df['behavioral_score']         = df[behav].sum(axis=1, min_count=1)
    df['behavioral_missing_count'] = df[behav].isna().sum(axis=1)

    opin = ['opinion_seas_vacc_effective','opinion_seas_risk','opinion_seas_sick_from_vacc']
    df['opinion_mean'] = df[opin].mean(axis=1)
    df['vacc_intent']  = df['opinion_seas_vacc_effective'] - df['opinion_seas_sick_from_vacc']

    df['risk_x_concern']            = df['opinion_seas_risk'] * df['flu_concern']
    df['effectiveness_x_knowledge'] = df['opinion_seas_vacc_effective'] * df['flu_knowledge']
    df['doc_rec_x_concern']         = df['doctor_recc_seasonal'] * df['flu_concern']
    df['household_total']           = df['household_adults'] + df['household_children']
    return df

train_fe = engineer(train).drop(columns=['age_group','education','income_poverty'])
test_fe  = engineer(test).drop(columns=['age_group','education','income_poverty'])

# %% [markdown]
# ## Split strategy — three disjoint sets
# - `dev_model`   : 68% of data — CV, tuning, fitting
# - `dev_explain` : 12% of data — XAI only (out-of-sample)
# - `holdout`     : 20% of data — final evaluation, **used once**

# %%
feature_cols = [c for c in train_fe.columns if c not in (TARGET_COL, ID_COL)]
X = train_fe[feature_cols].copy()
y = train_fe[TARGET_COL].copy()
X_test = test_fe[feature_cols].copy()

X_dev, X_hold, y_dev, y_hold = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y
)
X_dev_model, X_dev_explain, y_dev_model, y_dev_explain = train_test_split(
    X_dev, y_dev, test_size=0.15, random_state=RANDOM_STATE, stratify=y_dev
)
print('dev_model  :', X_dev_model.shape)
print('dev_explain:', X_dev_explain.shape)
print('holdout    :', X_hold.shape)

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

# %%
# ---------- PREPROCESSOR ----------
def make_preprocessor(X_ref):
    num_cols = X_ref.select_dtypes(include=[np.number]).columns.tolist()
    cat_cols = X_ref.select_dtypes(include=['object','category']).columns.tolist()
    return ColumnTransformer([
        ('num', Pipeline([('imputer', SimpleImputer(strategy='median')),
                          ('scaler', StandardScaler())]), num_cols),
        ('cat', Pipeline([
            ('imputer', SimpleImputer(strategy='constant', fill_value='MISSING')),
            ('ohe', OneHotEncoder(handle_unknown='ignore', sparse_output=False, min_frequency=20)),
        ]), cat_cols),
    ], remainder='drop')

# %%
def get_models():
    return {
        'lgbm': LGBMClassifier(n_estimators=600, learning_rate=0.05, num_leaves=31,
                               min_child_samples=20, subsample=0.9, colsample_bytree=0.9,
                               random_state=RANDOM_STATE, n_jobs=-1, verbose=-1),
        'xgb':  XGBClassifier(n_estimators=600, learning_rate=0.05, max_depth=5,
                              subsample=0.9, colsample_bytree=0.9, reg_lambda=1.0,
                              random_state=RANDOM_STATE, n_jobs=-1, eval_metric='auc'),
        'hist': HistGradientBoostingClassifier(max_iter=400, learning_rate=0.05,
                                               max_leaf_nodes=31, min_samples_leaf=20,
                                               l2_regularization=1.0, random_state=RANDOM_STATE),
        'rf':   RandomForestClassifier(n_estimators=600, min_samples_leaf=3,
                                       random_state=RANDOM_STATE, n_jobs=-1),
        'et':   ExtraTreesClassifier(n_estimators=600, min_samples_leaf=3,
                                     random_state=RANDOM_STATE, n_jobs=-1),
    }

# %% [markdown]
# ## Experiment: original vs engineered features (CV, cost-primary)
# Both runs use `dev_model` and CV. `age_group_ord` is the only age variable seen by any model.

# %%
raw_cols = [c for c in train.columns if c not in (TARGET_COL, ID_COL)]
X_raw_full = train[raw_cols].copy()
X_raw_full['age_group_ord'] = X_raw_full['age_group'].map(AGE_MAP)
X_raw_dev = X_raw_full.loc[X_dev_model.index].copy()

def cv_cost_auc(X_use, y_use, label):
    rows = []
    for name, model in get_models().items():
        pipe = Pipeline([('prep', make_preprocessor(X_use)), ('model', model)])
        res = cross_validate(pipe, X_use, y_use, cv=cv,
                             scoring={'cost': negative_mean_cost, 'auc': 'roc_auc'},
                             n_jobs=-1, error_score='raise')
        rows.append({'feature_set': label, 'model': name,
                     'cv_cost': -res['test_cost'].mean(),
                     'cv_auc':  res['test_auc'].mean()})
    return pd.DataFrame(rows)

fe_compare = pd.concat([
    cv_cost_auc(X_raw_dev, y_dev_model, 'original'),
    cv_cost_auc(X_dev_model, y_dev_model, 'engineered'),
])
display(fe_compare.sort_values(['feature_set', 'cv_cost']))

# %% [markdown]
# ## CV comparison on engineered features (primary metric = cost)

# %%
cv_rows = []
for name, model in get_models().items():
    pipe = Pipeline([('prep', make_preprocessor(X_dev_model)), ('model', model)])
    res = cross_validate(pipe, X_dev_model, y_dev_model, cv=cv,
                         scoring={'cost': negative_mean_cost, 'auc': 'roc_auc'},
                         n_jobs=-1, error_score='raise')
    cv_rows.append({'model': name,
                    'cv_cost_mean': -res['test_cost'].mean(),
                    'cv_cost_std' : res['test_cost'].std(),
                    'cv_auc_mean' : res['test_auc'].mean()})
    print(f"{name:5s} cost=${-res['test_cost'].mean():,.0f} | AUC {res['test_auc'].mean():.4f}")

cv_df = pd.DataFrame(cv_rows).sort_values('cv_cost_mean')
display(cv_df)

# %%
# ---------- TUNE TOP 3 BY COST ----------
param_grids = {
    'lgbm': {'model__n_estimators':[400,600,900],'model__learning_rate':[0.02,0.05,0.1],
             'model__num_leaves':[15,31,63],'model__min_child_samples':[10,20,40]},
    'xgb':  {'model__n_estimators':[400,600,900],'model__learning_rate':[0.02,0.05,0.1],
             'model__max_depth':[3,5,7],'model__reg_lambda':[0.1,1.0,5.0]},
    'hist': {'model__max_iter':[300,400,600],'model__learning_rate':[0.02,0.05,0.1],
             'model__max_leaf_nodes':[15,31,63]},
    'rf':   {'model__n_estimators':[400,600],'model__min_samples_leaf':[1,3,5],
             'model__max_depth':[None,15,25]},
    'et':   {'model__n_estimators':[400,600],'model__min_samples_leaf':[1,3,5],
             'model__max_depth':[None,15,25]},
}

tuned = []
for name in cv_df['model'].head(3).tolist():
    pipe = Pipeline([('prep', make_preprocessor(X_dev_model)), ('model', get_models()[name])])
    search = RandomizedSearchCV(pipe, param_grids[name], n_iter=10, cv=cv,
                                scoring=negative_mean_cost, n_jobs=-1,
                                random_state=RANDOM_STATE, refit=True)
    search.fit(X_dev_model, y_dev_model)
    tuned.append({'family': name, 'cv_cost': -search.best_score_,
                  'estimator': search.best_estimator_})
    print(f"Tuned {name}: cost=${-search.best_score_:,.0f}")

# %%
# ---------- ONE BEST PER FAMILY, RANKED BY CV COST (never by holdout) ----------
candidates, seen = [], set()
for t in sorted(tuned, key=lambda r: r['cv_cost']):
    if t['family'] in seen: continue
    candidates.append({'name': f'tuned_{t["family"]}', 'family': t['family'],
                       'cv_cost': t['cv_cost'], 'estimator': t['estimator']})
    seen.add(t['family'])

for _, row in cv_df.iterrows():
    if row['model'] in seen: continue
    pipe = Pipeline([('prep', make_preprocessor(X_dev_model)),
                     ('model', get_models()[row['model']])])
    pipe.fit(X_dev_model, y_dev_model)
    candidates.append({'name': f'untuned_{row["model"]}', 'family': row['model'],
                       'cv_cost': float(row['cv_cost_mean']), 'estimator': pipe})
    seen.add(row['model'])

candidates = sorted(candidates, key=lambda r: r['cv_cost'])
top5 = candidates[:5]
assert len(top5) == 5, f"Need 5 families, have {len(top5)}"
print('Selected top 5 by CV cost:')
for i, c in enumerate(top5, 1):
    print(f"  {i}. {c['name']:30s} cv_cost=${c['cv_cost']:,.0f}")

# %% [markdown]
# ## XAI on `dev_explain` (out-of-sample, holdout untouched)

# %%
try:
    import shap
    SHAP_OK = True
except Exception:
    SHAP_OK = False
    print('SHAP unavailable')

importance_store = {}
for c in top5:
    print(f'\n=== XAI: {c["name"]} ===')
    perm = permutation_importance(c['estimator'], X_dev_explain, y_dev_explain,
                                  n_repeats=5, random_state=RANDOM_STATE,
                                  scoring='roc_auc', n_jobs=-1)
    perm_df = pd.DataFrame({'feature': X_dev_explain.columns,
                            'importance_mean': perm.importances_mean,
                            'importance_std' : perm.importances_std}) \
                .sort_values('importance_mean', ascending=False)
    importance_store[c['name']] = perm_df.set_index('feature')['importance_mean']
    display(perm_df.head(10))
    plt.figure(figsize=(8,5))
    sns.barplot(data=perm_df.head(10), x='importance_mean', y='feature')
    plt.title(f'Permutation importance — {c["name"]}'); plt.tight_layout(); plt.show()

    if SHAP_OK:
        model = c['estimator'].named_steps['model']
        try:
            prep = c['estimator'].named_steps['prep']
            Xt = prep.transform(X_dev_explain)
            fnames = prep.get_feature_names_out()
            Xt_df = pd.DataFrame(Xt, columns=fnames)
            samp = Xt_df.sample(min(400, len(Xt_df)), random_state=RANDOM_STATE)
            explainer = shap.TreeExplainer(model)
            sv = explainer.shap_values(samp)
            if isinstance(sv, list):
                sv = sv[1] if len(sv) == 2 else sv[0]
            shap.summary_plot(sv, samp, show=False)
            plt.title(f'SHAP — {c["name"]}'); plt.tight_layout(); plt.show()
        except Exception as e:
            print('SHAP skipped:', e)

if importance_store:
    imp_df = pd.DataFrame(importance_store)
    plt.figure(figsize=(max(8, 0.6*len(imp_df)), max(6, 0.4*len(imp_df))))
    sns.heatmap(imp_df.fillna(0).sort_values(imp_df.columns[0], ascending=False),
                cmap='viridis')
    plt.title('Permutation importance across models'); plt.tight_layout(); plt.show()

# %%
# PDP on top numeric features of the best model (dev_explain, out-of-sample)
top_feats = importance_store[top5[0]['name']].head(2).index.tolist()
num_top = [f for f in top_feats if f in X_dev_explain.select_dtypes(include=[np.number]).columns]
if num_top:
    try:
        fig, ax = plt.subplots(1, len(num_top), figsize=(5*len(num_top), 4), squeeze=False)
        PartialDependenceDisplay.from_estimator(
            top5[0]['estimator'], X_dev_explain,
            features=num_top, ax=ax[0], grid_resolution=25)
        plt.suptitle('PDP — top model'); plt.tight_layout(); plt.show()
    except Exception as e:
        print('PDP skipped:', e)

# %% [markdown]
# ## Calibration on `dev_explain`

# %%
for c in top5:
    proba = c['estimator'].predict_proba(X_dev_explain)[:, 1]
    bs = brier_score_loss(y_dev_explain, proba)
    frac_pos, mean_pred = calibration_curve(y_dev_explain, proba, n_bins=10)
    plt.figure(figsize=(5,5))
    plt.plot(mean_pred, frac_pos, marker='o', label=f'{c["name"]}  Brier={bs:.4f}')
    plt.plot([0,1],[0,1],'--', color='grey')
    plt.xlabel('Predicted probability'); plt.ylabel('Observed frequency')
    plt.title(f'Reliability — {c["name"]}'); plt.legend(); plt.tight_layout(); plt.show()

# %% [markdown]
# ## Holdout evaluation — once only, for reporting

# %%
hold_rows = []
for c in top5:
    proba = c['estimator'].predict_proba(X_hold)[:, 1]
    c['holdout_auc']  = roc_auc_score(y_hold, proba)
    c['holdout_cost'] = total_cost(y_hold, proba, X_hold)
    hold_rows.append({'name': c['name'], 'family': c['family'],
                      'cv_cost': c['cv_cost'],
                      'holdout_cost': c['holdout_cost'],
                      'holdout_auc': c['holdout_auc']})
display(pd.DataFrame(hold_rows))

# Sensitivity on the top model
print('\nSensitivity on top model (holdout, reported once):')
proba_top = top5[0]['estimator'].predict_proba(X_hold)[:, 1]
for label, (hu, ve) in SENSITIVITY.items():
    print(f'  {label:8s} cost=${total_cost(y_hold, proba_top, X_hold, hu, ve):,.0f}')

# %%
# ---------- SUBMISSIONS: ranked by CV cost, not holdout ----------
SUBMIT_PROBABILITIES = True    # TODO: confirm with lecturer before uploading

files, preds_store = [], []
for rank, c in enumerate(top5, start=1):
    final = clone(c['estimator']).fit(X, y)   # final refit on full training data
    proba = final.predict_proba(X_test)[:, 1]

    if SUBMIT_PROBABILITIES:
        values = proba
    else:
        th = X_test['age_group_ord'].map(lambda o: threshold_for_ord(int(o), HU_CENTRAL, VE_CENTRAL)).values
        values = (proba >= th).astype(int)

    sub = pd.DataFrame({ID_COL: test[ID_COL].values, TARGET_COL: values})
    fname = f'challenge_submission_group_{GROUP_NUMBER}_order_{rank}.csv'
    sub.to_csv(fname, index=False)
    files.append({'rank': rank, 'file': fname, 'name': c['name'],
                  'cv_cost': c['cv_cost'],
                  'holdout_cost': c['holdout_cost'],
                  'holdout_auc': c['holdout_auc']})
    preds_store.append(np.asarray(values).copy())
    print(f"Saved {fname} | cv=${c['cv_cost']:,.0f} | holdout=${c['holdout_cost']:,.0f}")

# ---------- Safeguards ----------
for s in files:
    df = pd.read_csv(s['file'])
    assert df.shape == sample.shape, f"{s['file']} shape mismatch"
    assert list(df.columns) == list(sample.columns), f"{s['file']} columns mismatch"
    assert df[ID_COL].equals(sample[ID_COL]), \
        f"{s['file']} respondent IDs/order do not match the sample submission"
    assert not df[TARGET_COL].isna().any(), f"{s['file']} has NaN predictions"

if SUBMIT_PROBABILITIES:
    for s in files:
        df = pd.read_csv(s['file'])
        assert df[TARGET_COL].between(0, 1).all(), f"{s['file']} outside [0,1]"
else:
    for s in files:
        df = pd.read_csv(s['file'])
        assert set(np.unique(df[TARGET_COL])).issubset({0, 1}), \
            f"{s['file']} contains values other than 0/1"

for i in range(5):
    for j in range(i + 1, 5):
        assert not np.allclose(preds_store[i], preds_store[j]), \
            f"Submissions {i+1} and {j+1} are identical after formatting"

cv_seq = [s['cv_cost'] for s in files]
assert all(cv_seq[i] <= cv_seq[i + 1] for i in range(4)), \
    "Files not ordered best→worst by CV cost"

print('\nAll submission safeguards passed.')

# %%
summary = pd.DataFrame(files)
display(summary)
summary.to_csv('results_summary.csv', index=False)

with open('submission_links.txt', 'w') as f:
    f.write(f'Data Science Challenge — Town A (Seasonal Flu)\n')
    f.write(f'Group number: {GROUP_NUMBER}\n')
    f.write(f'Format: {"probabilities" if SUBMIT_PROBABILITIES else "binary"}\n\n')
    f.write('Code links (edit before submitting):\n')
    f.write('  GitHub Organisation: TODO\n')
    f.write('  Shared Colab Notebook (Editor access): TODO\n\n')
    f.write('Files (ranked best→worst by CV cost):\n')
    for s in files:
        f.write(f"  {s['file']}   cv=${s['cv_cost']:,.0f}   "
                f"holdout=${s['holdout_cost']:,.0f}   AUC={s['holdout_auc']:.4f}\n")
print('Wrote results_summary.csv and submission_links.txt')
```
