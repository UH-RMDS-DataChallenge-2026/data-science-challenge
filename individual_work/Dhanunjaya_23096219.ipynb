```python
# =============================================================================
# TOWN A – SEASONAL/H3N2 FLU – FINAL CORRECTED PIPELINE
# All Round-1 + Round-2 fixes applied (incl. local-file loading to fix HTTP 404)
# =============================================================================

# ============================================================
# [SECTION 0] CONFIG + IMPORTS
# ============================================================
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
pd.set_option("display.max_columns", 200)
pd.set_option("display.width", 200)

from sklearn.model_selection import (StratifiedKFold, RandomizedSearchCV,
                                     cross_val_predict, train_test_split)
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer, KNNImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import (RandomForestClassifier, ExtraTreesClassifier,
                              HistGradientBoostingClassifier)
from sklearn.metrics import (accuracy_score, roc_auc_score, log_loss,
                             brier_score_loss, confusion_matrix,
                             precision_score, recall_score, f1_score)
from sklearn.inspection import permutation_importance, PartialDependenceDisplay
from sklearn.calibration import calibration_curve
from sklearn.base import clone

sns.set_theme(style="whitegrid")
np.random.seed(42)

try:
    from xgboost import XGBClassifier
    XGB_AVAILABLE = True
except Exception:
    XGB_AVAILABLE = False

try:
    from lightgbm import LGBMClassifier
    LGBM_AVAILABLE = True
except Exception:
    LGBM_AVAILABLE = False

try:
    import shap
    SHAP_AVAILABLE = True
except Exception:
    SHAP_AVAILABLE = False

# --- [FIX-6 R2 / EDIT BEFORE RUNNING] ---
GROUP_NUMBER = 0                     # <-- CHANGE TO YOUR GROUP NUMBER
TARGET  = "seasonal_vaccine"
ID_COL  = "respondent_id"

VACCINE_COST  = 300
HOSPITAL_COST = 18_500

# --- [FIX-3 R2] CORRECTED hospitalisation & VE sourcing (see comments in report) ---
# CDC FluSurv-NET 2017-18 (H3N2): hosp/symptomatic illness 18-49:0.56%, 50-64:1.06%, 65+:9.09%
# Survey bins: 18-34/35-44 -> 0.56%; 45-54 -> 0.80% (INTERPOLATED, not CDC-reported);
# 55-64 -> 1.06%; 65+ -> 9.00%.  VE(H3N2 hosp) central 0.20 (0.10-0.30), NOT 41-61%.
VE_CENTRAL       = 0.20
R65_CENTRAL      = 0.09
CHRONIC_CENTRAL  = 1.0

RANDOM_STATE  = 42
N_FOLDS       = 5
INNER_FOLDS   = 3
HOLDOUT_FRAC  = 0.20
TUNE_ITER     = 15

# Example submission is 0.0 floats with only zeros -> binary expected.
SUBMISSION_FORMAT = "binary"   # "binary" or "probability"

# ============================================================
# [SECTION 1] LOAD DATA + VALIDATION  (FIXED: local-first loading)
# ============================================================
def _load_csv(local_name):
    """Load a CSV from local working dir; fall back to Colab upload prompt."""
    if os.path.exists(local_name):
        return pd.read_csv(local_name)
    try:
        from google.colab import files as _cf
        print(f"[INFO] Please upload: {local_name}")
        uploaded = _cf.upload()
        for name in uploaded:
            if name.endswith(".csv"):
                return pd.read_csv(name)
    except Exception:
        pass
    raise FileNotFoundError(
        f"Could not find '{local_name}'. Place it in the working directory."
    )

TRAIN_FILE   = "dataset_A_training.csv"
TEST_FILE    = "dataset_A_testing.csv"
EXAMPLE_FILE = "dataset_A_example_submission.csv"

train   = _load_csv(TRAIN_FILE)
test    = _load_csv(TEST_FILE)
example = _load_csv(EXAMPLE_FILE)

print(f"Train: {train.shape} | Test: {test.shape}")
print(f"Example columns: {list(example.columns)}")

# [FIX-11 R1] Data integrity checks
assert train[ID_COL].is_unique
assert test[ID_COL].is_unique
assert train[TARGET].notna().all()
assert set(train[TARGET].unique()).issubset({0, 1})
assert train.drop(columns=[TARGET, ID_COL]).columns.tolist() == \
       test.drop(columns=[ID_COL]).columns.tolist()
assert test[ID_COL].tolist() == example[ID_COL].tolist()

pred_col = [c for c in example.columns if c != ID_COL][0]

# [FIX-6 R2] Sanity-check the explicit format against the example
example_vals = set(example[pred_col].unique())
if SUBMISSION_FORMAT == "binary" and not example_vals.issubset({0, 1, 0.0, 1.0}):
    print(f"⚠ WARNING: example has non-binary values {list(example_vals)[:5]}. "
          f"Check if target should be probabilities.")
elif SUBMISSION_FORMAT == "probability" and example_vals.issubset({0, 1, 0.0, 1.0}):
    print(f"⚠ WARNING: example is 0/1 but SUBMISSION_FORMAT='probability'. "
          f"Confirm with lecturer before uploading.")
else:
    print(f"✓ SUBMISSION_FORMAT = '{SUBMISSION_FORMAT}' (consistent with example)")

# ============================================================
# [SECTION 2] NON-GRAPHICAL EDA
# ============================================================
print("\n=== Shape & dtypes ===")
print(f"Train {train.shape} | Test {test.shape}")
print(f"Features: {train.shape[1] - 2}")

missing_pct = train.isnull().mean().mul(100).sort_values(ascending=False)
print("\n--- Missingness (top 10) ---")
display(missing_pct[missing_pct > 0].head(10).round(2).to_frame("% missing"))

print("\n--- Target balance ---")
display(train[TARGET].value_counts(normalize=True).round(3).to_frame())

print("\n--- Numeric summary ---")
display(train.select_dtypes(include="number").describe().T.round(2))

print("\n--- Categorical cardinality ---")
for col in train.select_dtypes(include="object").columns:
    print(f"  {col}: {train[col].nunique()} unique")

print(f"\nDuplicate rows: {train.duplicated().sum()}")

# ============================================================
# [SECTION 3] GRAPHICAL EDA + ACTUAL FINDINGS
# ============================================================
if (missing_pct > 0).any():
    fig, ax = plt.subplots(figsize=(7, 6))
    missing_pct[missing_pct > 0].plot(kind="barh", color="tab:orange", ax=ax)
    ax.set_xlabel("% missing"); ax.set_title("Missing values per feature")
    plt.tight_layout(); plt.show()

print("FINDING – top-3 missing columns and their %:")
for col, pct in missing_pct[missing_pct > 0].head(3).items():
    print(f"  {col}: {pct:.1f}%")

key_feats = ["age_group", "doctor_recc_seasonal", "opinion_seas_risk",
             "opinion_seas_vacc_effective", "health_insurance",
             "number_chronic_med_conditions"]
key_feats = [f for f in key_feats if f in train.columns]

fig, axes = plt.subplots(2, 3, figsize=(16, 8))
for ax, f in zip(axes.ravel(), key_feats):
    train.groupby(f)[TARGET].mean().plot(kind="bar", ax=ax, color="tab:blue")
    ax.set_title(f"Uptake by {f}")
    ax.set_ylabel("Uptake rate")
    ax.tick_params(axis="x", rotation=45)
plt.tight_layout(); plt.show()

print("\nFINDING – actual uptake rates by feature:")
for f in key_feats:
    grp = train.groupby(f)[TARGET].agg(["mean", "count"]).round(3)
    print(f"\n--- {f} ---")
    display(grp)

num_feature_cols = (
    train.drop(columns=[TARGET, ID_COL], errors="ignore")
         .select_dtypes(include="number").columns
)
if len(num_feature_cols) > 1:
    plt.figure(figsize=(12, 10))
    sns.heatmap(train[num_feature_cols].corr(), cmap="coolwarm", center=0)
    plt.title("Correlation heatmap (features only, target excluded)")
    plt.tight_layout(); plt.show()

# ============================================================
# [SECTION 4] COST MODEL + HOSPITALISATION RISK
# ============================================================
def age_risk(age_label, r65=R65_CENTRAL):
    if pd.isna(age_label):
        return np.nan
    s = str(age_label).lower()
    if "65" in s:
        return r65
    if s.startswith("55"):
        return 0.0106
    if s.startswith("45"):
        return 0.008          # interpolation, documented above
    if s.startswith("35"):
        return 0.0056
    if s.startswith("18"):
        return 0.0056
    return np.nan             # unrecognised -> explicit NaN

def hosp_prob(df, r65=R65_CENTRAL, chronic_mult=CHRONIC_CENTRAL):
    base = (df["age_group"].map(lambda x: age_risk(x, r65))
            if "age_group" in df else pd.Series(np.nan, index=df.index))
    if base.isna().any():
        base = base.fillna(base.mean())      # mean-fill, NOT min
    chronic = (df["number_chronic_med_conditions"].fillna(0)
               if "number_chronic_med_conditions" in df else 0)
    return (base * (1 + chronic_mult * chronic)).clip(upper=1.0).to_numpy()

def total_cost(y_true, y_pred, h, ve=VE_CENTRAL):
    y_true, y_pred = np.asarray(y_true), np.asarray(y_pred)
    return np.where(
        y_pred == 1,
        VACCINE_COST + np.where(y_true == 1, h * (1 - ve), h) * HOSPITAL_COST,
        np.where(y_true == 1, VACCINE_COST, 0) + h * HOSPITAL_COST
    ).sum()

def cost_pp(y_true, y_pred, h, ve=VE_CENTRAL):
    return total_cost(y_true, y_pred, h, ve) / len(y_true)

def personal_threshold(h, ve=VE_CENTRAL):
    return VACCINE_COST / (VACCINE_COST + HOSPITAL_COST * h * ve)

# ============================================================
# [SECTION 5] DEV/HOLDOUT SPLIT + BASELINES
# ============================================================
train_dev, train_holdout = train_test_split(
    train, test_size=HOLDOUT_FRAC, stratify=train[TARGET],
    random_state=RANDOM_STATE)
train_dev     = train_dev.reset_index(drop=True)
train_holdout = train_holdout.reset_index(drop=True)

print(f"\nDev: {train_dev.shape} | Holdout: {train_holdout.shape}")

X_dev     = train_dev.drop(columns=[TARGET, ID_COL])
y_dev     = train_dev[TARGET].astype(int)
h_dev     = hosp_prob(train_dev)

X_holdout = train_holdout.drop(columns=[TARGET, ID_COL])
y_holdout = train_holdout[TARGET].astype(int)
h_holdout = hosp_prob(train_holdout)

X_full = train.drop(columns=[TARGET, ID_COL])
y_full = train[TARGET].astype(int)
X_test = test.drop(columns=[ID_COL])
h_test = hosp_prob(test)

def baseline_costs(y, h):
    y = np.asarray(y); n = len(y)
    return {
        "nobody":    total_cost(y, np.zeros(n), h) / n,
        "everybody": total_cost(y, np.ones(n),  h) / n,
        "perfect":   total_cost(y, y,           h) / n,
    }

baselines_dev     = baseline_costs(y_dev,     h_dev)
baselines_holdout = baseline_costs(y_holdout, h_holdout)

print("\n--- Baselines (dev) ---")
display(pd.Series(baselines_dev).round(2).to_frame("$/person"))
print("\n--- Baselines (holdout) ---")
display(pd.Series(baselines_holdout).round(2).to_frame("$/person"))

# ============================================================
# [SECTION 6] PRE-PROCESSING
# ============================================================
cat_cols = X_dev.select_dtypes(include="object").columns.tolist()
num_cols = X_dev.select_dtypes(exclude="object").columns.tolist()
OPINION_COLS = [c for c in ["opinion_seas_vacc_effective", "opinion_seas_risk",
                             "opinion_seas_sick_from_vacc"] if c in num_cols]

def _onehot():
    try:
        return OneHotEncoder(handle_unknown="ignore", sparse_output=False)
    except TypeError:
        return OneHotEncoder(handle_unknown="ignore", sparse=False)

def make_preprocessor(num_imputer="median", scale=False,
                      onehot_opinions=False, missing_indicator=True):
    num_used = [c for c in num_cols if not (onehot_opinions and c in OPINION_COLS)]
    imputer = (KNNImputer(n_neighbors=5, add_indicator=missing_indicator)
               if num_imputer == "knn"
               else SimpleImputer(strategy="median", add_indicator=missing_indicator))
    num_steps = [("impute", imputer)]
    if scale:
        num_steps.append(("scale", StandardScaler()))
    cat_pipe = Pipeline([
        ("impute", SimpleImputer(strategy="constant", fill_value="Missing")),
        ("onehot", _onehot()),
    ])
    parts = [("num", Pipeline(num_steps), num_used),
             ("cat", cat_pipe, cat_cols)]
    if onehot_opinions:
        op_pipe = Pipeline([
            ("impute", SimpleImputer(strategy="constant", fill_value=-1)),
            ("onehot", _onehot()),
        ])
        parts.append(("opinions", op_pipe, OPINION_COLS))
    return ColumnTransformer(parts)

# ============================================================
# [SECTION 7] BASE MODELS
# ============================================================
base_models = {
    "LogReg_median": Pipeline([
        ("prep", make_preprocessor("median", scale=True)),
        ("clf", LogisticRegression(max_iter=3000, random_state=RANDOM_STATE))]),
    "LogReg_KNN": Pipeline([
        ("prep", make_preprocessor("knn", scale=True)),
        ("clf", LogisticRegression(max_iter=3000, random_state=RANDOM_STATE))]),
    "RF": Pipeline([
        ("prep", make_preprocessor("median", scale=False)),
        ("clf", RandomForestClassifier(n_estimators=300, min_samples_leaf=5,
                                        n_jobs=-1, random_state=RANDOM_STATE))]),
    "ET": Pipeline([
        ("prep", make_preprocessor("median", scale=False)),
        ("clf", ExtraTreesClassifier(n_estimators=300, min_samples_leaf=5,
                                      n_jobs=-1, random_state=RANDOM_STATE))]),
    "Hist": Pipeline([
        ("prep", make_preprocessor("median", scale=False)),
        ("clf", HistGradientBoostingClassifier(random_state=RANDOM_STATE))]),
}
if XGB_AVAILABLE:
    base_models["XGB"] = Pipeline([
        ("prep", make_preprocessor("median", scale=False)),
        ("clf", XGBClassifier(n_estimators=300, learning_rate=0.05, max_depth=4,
                               subsample=0.9, colsample_bytree=0.9,
                               eval_metric="logloss", random_state=RANDOM_STATE,
                               n_jobs=-1))])
if LGBM_AVAILABLE:
    base_models["LGBM"] = Pipeline([
        ("prep", make_preprocessor("median", scale=False)),
        ("clf", LGBMClassifier(n_estimators=300, learning_rate=0.05,
                                num_leaves=31, random_state=RANDOM_STATE,
                                n_jobs=-1))])

print(f"\nActual base model count: {len(base_models)}")
print(f"Base models: {list(base_models.keys())}")

# ============================================================
# [SECTION 8] CV EVALUATION – [FIX-2 R2] unified predictions
# ============================================================
skf      = StratifiedKFold(n_splits=N_FOLDS,     shuffle=True, random_state=RANDOM_STATE)
inner_cv = StratifiedKFold(n_splits=INNER_FOLDS, shuffle=True, random_state=RANDOM_STATE)
GRID     = np.linspace(0.05, 0.95, 91)

def evaluate_cv(model, X, y, h, cv_outer=skf, grid=GRID):
    y_arr = np.asarray(y); n = len(y_arr)
    oof           = np.zeros(n)
    pred_05       = np.zeros(n, dtype=int)
    pred_global   = np.zeros(n, dtype=int)
    pred_personal = np.zeros(n, dtype=int)

    fold_thresholds  = []
    fold_costs_05    = []
    fold_costs_global = []
    fold_costs_personal = []

    for tr_idx, val_idx in cv_outer.split(X, y):
        X_tr, X_val = X.iloc[tr_idx], X.iloc[val_idx]
        y_tr, y_val = y.iloc[tr_idx], y.iloc[val_idx]
        h_tr, h_val = h[tr_idx], h[val_idx]

        inner_p = cross_val_predict(clone(model), X_tr, y_tr, cv=inner_cv,
                                     method="predict_proba", n_jobs=-1)[:, 1]
        inner_costs = [cost_pp(y_tr.values, (inner_p > t).astype(int), h_tr)
                        for t in grid]
        best_t = float(grid[np.argmin(inner_costs)])
        fold_thresholds.append(best_t)

        fitted = clone(model).fit(X_tr, y_tr)
        p_val  = fitted.predict_proba(X_val)[:, 1]
        oof[val_idx] = p_val

        pred_05[val_idx]       = (p_val > 0.5).astype(int)
        pred_global[val_idx]   = (p_val > best_t).astype(int)
        pred_personal[val_idx] = (p_val > personal_threshold(h_val)).astype(int)

        fold_costs_05.append(       cost_pp(y_val.values, pred_05[val_idx],       h_val))
        fold_costs_global.append(   cost_pp(y_val.values, pred_global[val_idx],   h_val))
        fold_costs_personal.append( cost_pp(y_val.values, pred_personal[val_idx], h_val))

    def metrics_for(pred):
        tn, fp, fn, tp = confusion_matrix(y_arr, pred).ravel()
        return {
            "cost_pp":   cost_pp(y_arr, pred, h),
            "accuracy":  accuracy_score(y_arr, pred),
            "precision": precision_score(y_arr, pred, zero_division=0),
            "recall":    recall_score(y_arr, pred, zero_division=0),
            "f1":        f1_score(y_arr, pred, zero_division=0),
            "send_rate": float(pred.mean()),
            "TN": int(tn), "FP": int(fp), "FN": int(fn), "TP": int(tp),
        }

    return {
        "oof_preds": oof,
        "frozen_global_threshold": float(np.median(fold_thresholds)),
        "strategies": {
            "personal": metrics_for(pred_personal),
            "0.5":      metrics_for(pred_05),
            "global_t": metrics_for(pred_global),
        },
        "cost_personal_sd": np.std(fold_costs_personal),
        "cost_05_sd":       np.std(fold_costs_05),
        "cost_global_sd":   np.std(fold_costs_global),
        "auc":     roc_auc_score(y_arr, oof),
        "logloss": log_loss(y_arr, oof),
        "brier":   brier_score_loss(y_arr, oof),
    }

# ============================================================
# [SECTION 9] HYPERPARAMETER TUNING
# [FIX-1 R2] DISCLOSURE: hyperparameters are selected using all of X_dev.
# CV costs for TUNED models are development/model-selection results, not unbiased
# outer-CV estimates. Holdout (§11) is the only unbiased arbiter.
# ============================================================
param_grids = {
    "RF": {
        "clf__n_estimators":     [200, 400, 600],
        "clf__min_samples_leaf": [1, 3, 5, 10],
        "clf__max_features":     ["sqrt", "log2", None],
        "clf__max_depth":        [None, 10, 20],
    },
    "ET": {
        "clf__n_estimators":     [200, 400, 600],
        "clf__min_samples_leaf": [1, 3, 5, 10],
        "clf__max_features":     ["sqrt", "log2", None],
        "clf__max_depth":        [None, 10, 20],
    },
    "Hist": {
        "clf__learning_rate":     [0.02, 0.05, 0.1],
        "clf__max_depth":         [None, 3, 5, 8],
        "clf__max_leaf_nodes":    [15, 31, 63],
        "clf__min_samples_leaf":  [20, 50, 100],
        "clf__l2_regularization": [0.0, 0.1, 1.0],
    },
}
if XGB_AVAILABLE:
    param_grids["XGB"] = {
        "clf__n_estimators":     [200, 400, 600],
        "clf__learning_rate":    [0.02, 0.05, 0.1],
        "clf__max_depth":        [3, 4, 6],
        "clf__subsample":        [0.7, 0.9, 1.0],
        "clf__colsample_bytree": [0.7, 0.9, 1.0],
    }
if LGBM_AVAILABLE:
    param_grids["LGBM"] = {
        "clf__n_estimators":      [200, 400, 600],
        "clf__learning_rate":     [0.02, 0.05, 0.1],
        "clf__num_leaves":        [15, 31, 63],
        "clf__min_child_samples": [10, 20, 50],
    }

def cost_scorer(estimator, X, y):
    h = hosp_prob(X)
    p = estimator.predict_proba(X)[:, 1]
    pred = (p > personal_threshold(h)).astype(int)
    return -total_cost(np.asarray(y), pred, h) / len(y)

print("\n=== Initial CV on dev (base models) ===")
cv_results_base = {}
for name, model in base_models.items():
    print(f"  CV: {name}")
    cv_results_base[name] = evaluate_cv(model, X_dev, y_dev, h_dev)

initial_rank = pd.DataFrame([
    {"model": n,
     "cv_cost_personal": r["strategies"]["personal"]["cost_pp"],
     "cv_cost_0_5":      r["strategies"]["0.5"]["cost_pp"],
     "cv_cost_global":   r["strategies"]["global_t"]["cost_pp"],
     "auc":              r["auc"],
     "logloss":          r["logloss"],
     "brier":            r["brier"]}
    for n, r in cv_results_base.items()
]).sort_values("cv_cost_personal").reset_index(drop=True)

print("\nInitial ranking (by personal-threshold CV cost):")
display(initial_rank.round(3))

top3_unique = initial_rank["model"].head(3).tolist()
print(f"\nTuning top-3 models: {top3_unique}")

tuned_models = {}
tune_log = []
for name in top3_unique:
    if name not in param_grids:
        print(f"  {name}: no grid – skipped")
        continue

    search_ll = RandomizedSearchCV(
        base_models[name], param_grids[name], n_iter=TUNE_ITER,
        scoring="neg_log_loss", cv=inner_cv, n_jobs=-1,
        random_state=RANDOM_STATE, verbose=0)
    search_ll.fit(X_dev, y_dev)

    search_cost = RandomizedSearchCV(
        base_models[name], param_grids[name], n_iter=TUNE_ITER,
        scoring=cost_scorer, cv=inner_cv, n_jobs=-1,
        random_state=RANDOM_STATE, verbose=0)
    search_cost.fit(X_dev, y_dev)

    tuned_models[f"{name}_tuned_ll"]   = search_ll.best_estimator_
    tuned_models[f"{name}_tuned_cost"] = search_cost.best_estimator_

    tune_log.append({
        "model":       name,
        "LL_score":    round(-search_ll.best_score_, 4),
        "Cost_score":  round(-search_cost.best_score_, 2),
    })
    print(f"  {name}: LL = {-search_ll.best_score_:.4f} | Cost = {-search_cost.best_score_:.2f} $/person")

print("\n--- Tuning summary ---")
display(pd.DataFrame(tune_log))

# ============================================================
# [SECTION 10] FINAL CV RANKING + MODEL-FAMILY SELECTION
# ============================================================
all_models = {**base_models, **tuned_models}
print(f"\nTotal models evaluated: {len(all_models)}")

print("\n=== Final CV on dev (base + tuned) ===")
cv_results_all = {}
for name, model in all_models.items():
    print(f"  CV: {name}")
    cv_results_all[name] = evaluate_cv(model, X_dev, y_dev, h_dev)

def model_family(name):
    n = name.lower()
    if n.startswith("logreg"):  return "LogReg"
    if n.startswith("rf"):      return "RandomForest"
    if n.startswith("et"):      return "ExtraTrees"
    if n.startswith("hist"):    return "HistGradientBoosting"
    if n.startswith("xgb"):     return "XGBoost"
    if n.startswith("lgbm"):    return "LightGBM"
    return name

final_rows = []
for name, res in cv_results_all.items():
    for strat, sdict in res["strategies"].items():
        sd = {"personal": res["cost_personal_sd"],
              "0.5":      res["cost_05_sd"],
              "global_t": res["cost_global_sd"]}[strat]
        final_rows.append({
            "family":     model_family(name),
            "model":      name,
            "strategy":   strat,
            "threshold":  0.5 if strat == "0.5"
                          else res["frozen_global_threshold"] if strat == "global_t"
                          else np.nan,
            "cv_cost":    sdict["cost_pp"],
            "cv_cost_sd": sd,
            "accuracy":   sdict["accuracy"],
            "precision":  sdict["precision"],
            "recall":     sdict["recall"],
            "f1":         sdict["f1"],
            "auc":        res["auc"],
            "logloss":    res["logloss"],
            "brier":      res["brier"],
            "send_rate":  sdict["send_rate"],
            "TN": sdict["TN"], "FP": sdict["FP"],
            "FN": sdict["FN"], "TP": sdict["TP"],
        })

final_df = pd.DataFrame(final_rows)
final_df["saving_vs_nobody"]    = baselines_dev["nobody"]    - final_df["cv_cost"]
final_df["saving_vs_everybody"] = baselines_dev["everybody"] - final_df["cv_cost"]

final_df["cv_bias_status"] = final_df["model"].apply(
    lambda n: "selection-biased CV (disclosed §9)" if "tuned" in n
              else "unbiased outer-CV (nested threshold)"
)

print("\n=== DEV CV PERFORMANCE MATRIX (sorted by cost) ===")
perf_cols = ["family","model","strategy","threshold","cv_cost","cv_cost_sd",
             "accuracy","precision","recall","f1","auc","logloss","brier",
             "saving_vs_nobody","send_rate","cv_bias_status"]
display(final_df.sort_values("cv_cost")[perf_cols].head(20).round(4))

best_per_family = (
    final_df.sort_values("cv_cost")
            .drop_duplicates(subset=["family"], keep="first")
            .reset_index(drop=True)
)

print("\n=== Best strategy per FAMILY (dev CV) ===")
display(best_per_family[["family","model","strategy","cv_cost","cv_cost_sd",
                          "auc","saving_vs_nobody","send_rate"]].round(3))

candidates_df = best_per_family.head(5).reset_index(drop=True)
print(f"\n=== 5 FINAL CANDIDATES ({len(candidates_df)} distinct families) ===")
display(candidates_df[["family","model","strategy","cv_cost","auc"]].round(3))

# ============================================================
# [SECTION 11] HOLDOUT EVALUATION – frozen thresholds only
# ============================================================
print("\n=== HOLDOUT EVALUATION (frozen dev thresholds) ===")

holdout_rows = []
holdout_probs = {}
for name, res in cv_results_all.items():
    model = all_models[name]
    model.fit(X_dev, y_dev)
    p_hold = model.predict_proba(X_holdout)[:, 1]
    holdout_probs[name] = p_hold

    frozen_t = res["frozen_global_threshold"]
    for strat, pred in [
        ("personal", (p_hold > personal_threshold(h_holdout)).astype(int)),
        ("0.5",      (p_hold > 0.5).astype(int)),
        ("global_t", (p_hold > frozen_t).astype(int)),
    ]:
        tn, fp, fn, tp = confusion_matrix(y_holdout, pred).ravel()
        holdout_rows.append({
            "family":   model_family(name),
            "model":    name,
            "strategy": strat,
            "threshold": 0.5 if strat == "0.5"
                          else frozen_t if strat == "global_t"
                          else np.nan,
            "holdout_cost": cost_pp(y_holdout, pred, h_holdout),
            "accuracy":     accuracy_score(y_holdout, pred),
            "precision":    precision_score(y_holdout, pred, zero_division=0),
            "recall":       recall_score(y_holdout, pred, zero_division=0),
            "f1":           f1_score(y_holdout, pred, zero_division=0),
            "auc":          roc_auc_score(y_holdout, p_hold),
            "logloss":      log_loss(y_holdout, p_hold),
            "brier":        brier_score_loss(y_holdout, p_hold),
            "send_rate":    float(pred.mean()),
            "TN": int(tn), "FP": int(fp), "FN": int(fn), "TP": int(tp),
        })

holdout_df = pd.DataFrame(holdout_rows).sort_values("holdout_cost").reset_index(drop=True)
holdout_df["saving_vs_nobody"]    = baselines_holdout["nobody"]    - holdout_df["holdout_cost"]
holdout_df["saving_vs_everybody"] = baselines_holdout["everybody"] - holdout_df["holdout_cost"]

print("\n=== HOLDOUT PERFORMANCE MATRIX ===")
display(holdout_df[["family","model","strategy","holdout_cost","accuracy","auc",
                     "logloss","brier","saving_vs_nobody","send_rate"]].head(20).round(4))

print("\n=== Dev vs holdout consistency for the 5 candidates ===")
comp = []
for _, r in candidates_df.iterrows():
    hh = holdout_df[(holdout_df["model"] == r["model"]) &
                    (holdout_df["strategy"] == r["strategy"])]
    if len(hh):
        comp.append({
            "family": r["family"], "model": r["model"], "strategy": r["strategy"],
            "dev_cv_cost":  round(r["cv_cost"], 2),
            "holdout_cost": round(hh.iloc[0]["holdout_cost"], 2),
            "delta":        round(hh.iloc[0]["holdout_cost"] - r["cv_cost"], 2),
        })
display(pd.DataFrame(comp))

# ============================================================
# [SECTION 12] SENSITIVITY ANALYSIS (corrected VE)
# ============================================================
print("\n=== Sensitivity analysis (corrected VE = 0.20 central) ===")
def make_h(df, r65, chronic_mult):
    base = (df["age_group"].map(lambda x: age_risk(x, r65))
            if "age_group" in df else pd.Series(np.nan, index=df.index))
    if base.isna().any():
        base = base.fillna(base.mean())
    chronic = (df["number_chronic_med_conditions"].fillna(0)
               if "number_chronic_med_conditions" in df else 0)
    return (base * (1 + chronic_mult * chronic)).clip(upper=1.0).to_numpy()

scenarios = {
    "Central (VE=0.20, 65+=9%, chronic ×(1+n))": (0.20, 0.09, 1.0),
    "Lower VE=0.10":                              (0.10, 0.09, 1.0),
    "Upper VE=0.30":                              (0.30, 0.09, 1.0),
    "65+ risk = 2.5%":                            (0.20, 0.025, 1.0),
    "65+ risk = 5%":                              (0.20, 0.05, 1.0),
    "Weaker chronic ×(1+0.5n)":                   (0.20, 0.09, 0.5),
    "No chronic adjustment ×(1+0n)":              (0.20, 0.09, 0.0),
}

top_models_names = candidates_df["model"].tolist()
sens_rows = []
for sc, (ve_s, r65_s, cm_s) in scenarios.items():
    h_s = make_h(train_holdout, r65_s, cm_s)
    t_pers_s = VACCINE_COST / (VACCINE_COST + HOSPITAL_COST * h_s * ve_s)
    row = {
        "scenario":  sc,
        "nobody":    cost_pp(y_holdout, np.zeros(len(y_holdout)), h_s, ve_s),
        "everybody": cost_pp(y_holdout, np.ones(len(y_holdout)),  h_s, ve_s),
    }
    for m in top_models_names:
        p = holdout_probs[m]
        row[f"{m}|personal"] = cost_pp(y_holdout, (p > t_pers_s).astype(int), h_s, ve_s)
        row[f"{m}|0.5"]      = cost_pp(y_holdout, (p > 0.5).astype(int),    h_s, ve_s)
    sens_rows.append(row)

sens_df = pd.DataFrame(sens_rows).set_index("scenario")
print("Sensitivity matrix (holdout cost/person $):")
display(sens_df.round(1).T)

# ============================================================
# [SECTION 13] FINAL 5 SUBMISSIONS
# ============================================================
print("\n=== Building 5 final submissions ===")
submission_checks = []
for order, row in enumerate(candidates_df.itertuples(), start=1):
    name, strat = row.model, row.strategy
    model = all_models[name]

    model.fit(X_full, y_full)
    p_test = model.predict_proba(X_test)[:, 1]

    if strat == "0.5":
        t_test = 0.5
    elif strat == "global_t":
        t_test = cv_results_all[name]["frozen_global_threshold"]
    else:
        t_test = personal_threshold(h_test)

    if SUBMISSION_FORMAT == "binary":
        pred_vals = (p_test > t_test).astype(int)
    else:
        pred_vals = np.clip(p_test, 0, 1)

    sub = pd.DataFrame({ID_COL: test[ID_COL], pred_col: pred_vals})
    fname = f"challenge_submission_group_{GROUP_NUMBER}_order_{order}.csv"
    sub.to_csv(fname, index=False)

    submission_checks.append({
        "order":      order,
        "family":     row.family,
        "model":      name,
        "strategy":   strat,
        "dev_cv_cost": round(row.cv_cost, 2),
        "file":       fname,
        "rows":       len(sub),
        "ids_match":  list(sub[ID_COL]) == list(example[ID_COL]),
        "format_ok":  SUBMISSION_FORMAT == "probability"
                      or bool(sub[pred_col].isin([0, 1]).all()),
        "no_missing": bool(sub[pred_col].notna().all()),
        "send_rate":  round(float((pred_vals > 0.5).mean()
                                   if SUBMISSION_FORMAT == "probability"
                                   else pred_vals.mean()), 3),
    })

print("\n--- Submission validation ---")
display(pd.DataFrame(submission_checks))

files = [f"challenge_submission_group_{GROUP_NUMBER}_order_{o}.csv"
         for o in range(1, 6)]
dfs = [pd.read_csv(f)[pred_col].values for f in files]
print("\nUniqueness check (rows differing between pairs):")
for i in range(len(dfs)):
    for j in range(i + 1, len(dfs)):
        print(f"  order {i+1} vs order {j+1}: {int((dfs[i] != dfs[j]).sum())} differing rows")

try:
    from google.colab import files as colab_files
    for order in range(1, 6):
        colab_files.download(f"challenge_submission_group_{GROUP_NUMBER}_order_{order}.csv")
except Exception:
    pass

# ============================================================
# [SECTION 14] XAI FOR ALL SHORTLISTED MODELS
# ============================================================
print("\n=== XAI across the 5 candidates (on holdout sample) ===")

def cost_permutation_importance(model, X_s, y_s, h_s, n_repeats=5, rng_seed=42):
    """[FIX-8 R2] CAVEAT — h_s is held FIXED while features are shuffled.
    Measures model sensitivity, NOT causal effect of changing age/chronic status."""
    rng = np.random.default_rng(rng_seed)
    y_arr = y_s.values if hasattr(y_s, "values") else np.asarray(y_s)
    p = model.predict_proba(X_s)[:, 1]
    base_cost = cost_pp(y_arr, (p > personal_threshold(h_s)).astype(int), h_s)
    out = {}
    for col in X_s.columns:
        deltas = []
        for _ in range(n_repeats):
            Xp = X_s.copy()
            Xp[col] = rng.permutation(Xp[col].values)
            pp = model.predict_proba(Xp)[:, 1]
            deltas.append(cost_pp(y_arr,
                                   (pp > personal_threshold(h_s)).astype(int),
                                   h_s) - base_cost)
        out[col] = float(np.mean(deltas))
    return pd.Series(out).sort_values(ascending=False)

def get_shap_explainer_and_values(model, X_sample):
    """[FIX-5 R2] Handles 2D and 3D SHAP output; extracts positive class."""
    prep = model.named_steps["prep"]
    clf  = model.named_steps["clf"]
    X_enc = prep.transform(X_sample)
    X_enc = X_enc.toarray() if hasattr(X_enc, "toarray") else X_enc
    feat_names = prep.get_feature_names_out()
    X_enc_df = pd.DataFrame(X_enc, columns=feat_names)

    if isinstance(clf, LogisticRegression):
        explainer = shap.LinearExplainer(clf, X_enc_df)
    else:
        explainer = shap.TreeExplainer(clf)

    shap_vals = explainer(X_enc_df)

    values = shap_vals.values
    if values.ndim == 3:
        shap_vals.values = values[:, :, 1]        # positive class
    elif values.ndim == 1:
        shap_vals.values = values.reshape(-1, 1)

    return feat_names, X_enc_df, shap_vals

rng = np.random.default_rng(RANDOM_STATE)
xai_idx = rng.choice(len(X_holdout), size=min(1500, len(X_holdout)), replace=False)
X_xai = X_holdout.iloc[xai_idx]
y_xai = y_holdout.iloc[xai_idx]
h_xai = h_holdout[xai_idx]

xai_summary = {}
for name in candidates_df["model"]:
    print(f"\n--- XAI: {name} ---")
    model = all_models[name]
    model.fit(X_dev, y_dev)

    pi_auc = permutation_importance(model, X_xai, y_xai,
                                     scoring="roc_auc", n_repeats=5,
                                     random_state=RANDOM_STATE, n_jobs=-1)
    pi_df = (pd.DataFrame({"feature": X_xai.columns,
                           "imp_mean": pi_auc.importances_mean,
                           "imp_std":  pi_auc.importances_std})
               .sort_values("imp_mean", ascending=False))
    print("Top-10 AUC permutation importance:")
    display(pi_df.head(10).round(4))

    cpi = cost_permutation_importance(model, X_xai, y_xai, h_xai)
    print("Top-10 COST-based importance (extra $/person if shuffled):")
    display(cpi.head(10).to_frame("delta_cost_pp").round(2))

    p_hold_full = holdout_probs[name]
    pt, pp = calibration_curve(y_holdout, p_hold_full, n_bins=10, strategy="quantile")
    plt.figure(figsize=(5, 5))
    plt.plot(pp, pt, marker="o", label=name)
    plt.plot([0, 1], [0, 1], "--", color="grey")
    plt.xlabel("Predicted probability")
    plt.ylabel("Observed uptake")
    plt.title(f"Calibration – {name}")
    plt.legend(); plt.tight_layout(); plt.show()

    if SHAP_AVAILABLE:
        try:
            feat_names, X_enc_df, shap_vals = get_shap_explainer_and_values(model, X_xai)

            vals = shap_vals.values
            if vals.ndim == 3:
                vals = vals[:, :, 1]
            mean_abs = np.abs(vals).mean(axis=0)
            top_idx  = np.argsort(mean_abs)[-5:][::-1]
            top_features = [feat_names[i] for i in top_idx]
            print("Top-5 features by mean |SHAP|:", top_features)
            xai_summary[name] = {"top_shap": top_features}

            shap.plots.beeswarm(shap_vals, max_display=12, show=True)
            shap.plots.waterfall(shap_vals[0], show=True)
        except Exception as e:
            print(f"    SHAP skipped for {name}: {e}")

print("\n=== PDP for top-2 candidates ===")
for name in candidates_df["model"].head(2):
    model = all_models[name]; model.fit(X_dev, y_dev)
    pi = permutation_importance(model, X_xai, y_xai, scoring="roc_auc",
                                 n_repeats=3, random_state=RANDOM_STATE, n_jobs=-1)
    top_feat = X_xai.columns[np.argmax(pi.importances_mean)]
    print(f"  {name}: PDP for {top_feat}")
    try:
        PartialDependenceDisplay.from_estimator(model, X_xai, [top_feat],
                                                 kind="average")
        plt.tight_layout(); plt.show()
    except Exception as e:
        print(f"    PDP failed: {e}")

# ============================================================
# [SECTION 15] FINAL SUMMARY MATRIX
# ============================================================
print("\n=== FINAL SUMMARY – 5 SUBMISSIONS ===")
summary = candidates_df[["family","model","strategy","cv_cost","cv_cost_sd",
                          "accuracy","precision","recall","f1",
                          "auc","logloss","brier",
                          "saving_vs_nobody","send_rate"]].copy()
summary.insert(0, "rank", range(1, len(summary) + 1))
display(summary.round(4))

print("\n=== HOLDOUT VERIFICATION FOR THE 5 SUBMISSIONS ===")
verify = (holdout_df[holdout_df["model"].isin(candidates_df["model"])]
          [["family","model","strategy","holdout_cost","accuracy","auc",
            "saving_vs_nobody","send_rate"]]
          .round(3).reset_index(drop=True))
display(verify)

print("\n✅ PIPELINE COMPLETE")
print(f"   Submission format: {SUBMISSION_FORMAT}")
print(f"   Group number: {GROUP_NUMBER}")
print("\nFinal checklist before uploading:")
print("  □ BASE URL removed – CSVs loaded locally or via upload prompt")
print("  □ Set GROUP_NUMBER above")
print("  □ Push notebook to GitHub; ensure Colab link is Editor-accessible")
print("  □ Upload the 5 challenge_submission_group_X_order_Y.csv files to Canvas")
print("  □ Verify files are 0/1 (or probabilities if SUBMISSION_FORMAT='probability')")
print("  □ Verify files are non-identical (uniqueness check above)")
print("  □ Every team member has visible GitHub commits and presents in-class")
```
